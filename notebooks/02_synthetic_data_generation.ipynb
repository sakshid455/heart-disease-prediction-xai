````python
```xml
<!-- filepath: c:\Users\datir\predictive\notebooks\02_synthetic_data_generation.ipynb -->
<VSCode.Cell id="1" language="markdown">
# Synthetic Data Generation Notebook

This notebook demonstrates how to generate synthetic datasets for various machine learning tasks.
</VSCode.Cell>
<VSCode.Cell id="2" language="python">
# Import Required Libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import sys
import os

# Add src directory to path
sys.path.insert(0, os.path.join(os.getcwd(), '..', 'src'))

from synthetic_data import (
    generate_synthetic_dataset,
    generate_classification_data,
    generate_regression_data,
    add_noise
)

# Set visualization style
sns.set_style("whitegrid")
plt.rcParams['figure.figsize'] = (12, 6)
</VSCode.Cell>
<VSCode.Cell id="3" language="markdown">
## Classification Data Generation

Generate synthetic data for classification tasks.
</VSCode.Cell>
<VSCode.Cell id="4" language="python">
# Generate classification dataset
X_class, y_class = generate_classification_data(
    n_samples=1000,
    n_features=10,
    n_classes=3,
    random_state=42
)

print(f"Classification Dataset:")
print(f"  Features shape: {X_class.shape}")
print(f"  Target shape: {y_class.shape}")
print(f"  Class distribution: {np.bincount(y_class)}")

# Convert to DataFrame
df_class = pd.DataFrame(X_class, columns=[f'feature_{i}' for i in range(X_class.shape[1])])
df_class['target'] = y_class
print(f"\nDataFrame shape: {df_class.shape}")
print(df_class.head())
</VSCode.Cell>
<VSCode.Cell id="5" language="markdown">
## Regression Data Generation

Generate synthetic data for regression tasks.
</VSCode.Cell>
<VSCode.Cell id="6" language="python">
# Generate regression dataset
X_reg, y_reg = generate_regression_data(
    n_samples=1000,
    n_features=10,
    random_state=42
)

print(f"Regression Dataset:")
print(f"  Features shape: {X_reg.shape}")
print(f"  Target shape: {y_reg.shape}")
print(f"  Target statistics:")
print(f"    Mean: {y_reg.mean():.2f}")
print(f"    Std: {y_reg.std():.2f}")
print(f"    Min: {y_reg.min():.2f}")
print(f"    Max: {y_reg.max():.2f}")

# Convert to DataFrame
df_reg = pd.DataFrame(X_reg, columns=[f'feature_{i}' for i in range(X_reg.shape[1])])
df_reg['target'] = y_reg
print(f"\nDataFrame shape: {df_reg.shape}")
print(df_reg.head())
</VSCode.Cell>
<VSCode.Cell id="7" language="markdown">
## Adding Noise to Data

Generate datasets with added noise for robustness testing.
</VSCode.Cell>
<VSCode.Cell id="8" language="python">
# Create clean and noisy versions
df_clean = generate_synthetic_dataset(n_samples=500, n_features=5, task='classification')
df_noisy = add_noise(df_clean, noise_level=0.2, random_state=42)

print(f"Clean dataset shape: {df_clean.shape}")
print(f"Noisy dataset shape: {df_noisy.shape}")

# Compare feature statistics
print("\nFeature Statistics Comparison (feature_0):")
print(f"Clean - Mean: {df_clean['feature_0'].mean():.4f}, Std: {df_clean['feature_0'].std():.4f}")
print(f"Noisy - Mean: {df_noisy['feature_0'].mean():.4f}, Std: {df_noisy['feature_0'].std():.4f}")
</VSCode.Cell>
<VSCode.Cell id="9" language="python">
# Visualize clean vs noisy data
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Clean data
axes[0].scatter(df_clean['feature_0'], df_clean['feature_1'], c=df_clean['target'], cmap='viridis', alpha=0.6)
axes[0].set_title('Clean Data')
axes[0].set_xlabel('feature_0')
axes[0].set_ylabel('feature_1')

# Noisy data
axes[1].scatter(df_noisy['feature_0'], df_noisy['feature_1'], c=df_noisy['target'], cmap='viridis', alpha=0.6)
axes[1].set_title('Noisy Data')
axes[1].set_xlabel('feature_0')
axes[1].set_ylabel('feature_1')

plt.tight_layout()
plt.show()
</VSCode.Cell>
<VSCode.Cell id="10" language="markdown">
## Saving Generated Data

Save the generated datasets for use in model training.
</VSCode.Cell>
<VSCode.Cell id="11" language="python">
# Save datasets
data_dir = os.path.join(os.getcwd(), '..', 'data', 'processed')
os.makedirs(data_dir, exist_ok=True)

df_class.to_csv(os.path.join(data_dir, 'classification_data.csv'), index=False)
df_reg.to_csv(os.path.join(data_dir, 'regression_data.csv'), index=False)
df_clean.to_csv(os.path.join(data_dir, 'clean_data.csv'), index=False)
df_noisy.to_csv(os.path.join(data_dir, 'noisy_data.csv'), index=False)

print("Datasets saved successfully!")
print(f"  - classification_data.csv")
print(f"  - regression_data.csv")
print(f"  - clean_data.csv")
print(f"  - noisy_data.csv")
</VSCode.Cell>
```
````
