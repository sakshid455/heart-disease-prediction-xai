````python
```xml
<!-- filepath: c:\Users\datir\predictive\notebooks\01_data_analysis.ipynb -->
<VSCode.Cell id="1" language="markdown">
# Data Analysis Notebook

This notebook performs exploratory data analysis (EDA) on the dataset.
</VSCode.Cell>
<VSCode.Cell id="2" language="python">
# Import Required Libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import sys
import os

# Add src directory to path
sys.path.insert(0, os.path.join(os.getcwd(), '..', 'src'))

from data_preprocessing import load_data, handle_missing_values, remove_duplicates

# Set visualization style
sns.set_style("whitegrid")
plt.rcParams['figure.figsize'] = (12, 6)
</VSCode.Cell>
<VSCode.Cell id="3" language="markdown">
## Load Data

Load the dataset and display basic information.
</VSCode.Cell>
<VSCode.Cell id="4" language="python">
# Create sample data for demonstration
from synthetic_data import generate_synthetic_dataset

# Generate synthetic classification data
df = generate_synthetic_dataset(n_samples=500, n_features=5, task='classification')

print("Dataset Shape:", df.shape)
print("\nFirst 5 rows:")
print(df.head())
print("\nDataset Info:")
print(df.info())
print("\nBasic Statistics:")
print(df.describe())
</VSCode.Cell>
<VSCode.Cell id="5" language="markdown">
## Data Quality Assessment

Check for missing values, duplicates, and data types.
</VSCode.Cell>
<VSCode.Cell id="6" language="python">
# Check for missing values
print("Missing Values:")
print(df.isnull().sum())
print(f"\nTotal missing values: {df.isnull().sum().sum()}")

# Check for duplicates
print(f"\nDuplicate rows: {df.duplicated().sum()}")

# Data types
print("\nData Types:")
print(df.dtypes)
</VSCode.Cell>
<VSCode.Cell id="7" language="markdown">
## Exploratory Data Analysis

Analyze distributions and relationships in the data.
</VSCode.Cell>
<VSCode.Cell id="8" language="python">
# Feature distributions
fig, axes = plt.subplots(2, 3, figsize=(15, 8))
axes = axes.ravel()

for idx, col in enumerate(df.columns[:-1]):
    axes[idx].hist(df[col], bins=30, edgecolor='black', alpha=0.7)
    axes[idx].set_title(f'Distribution of {col}')
    axes[idx].set_xlabel(col)
    axes[idx].set_ylabel('Frequency')

# Remove extra subplot
if len(df.columns) < 6:
    fig.delaxes(axes[-1])

plt.tight_layout()
plt.show()

print("Distribution analysis complete")
</VSCode.Cell>
<VSCode.Cell id="9" language="python">
# Correlation analysis
numeric_cols = df.select_dtypes(include=[np.number]).columns
correlation_matrix = df[numeric_cols].corr()

plt.figure(figsize=(10, 8))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', center=0)
plt.title('Correlation Matrix')
plt.tight_layout()
plt.show()
</VSCode.Cell>
<VSCode.Cell id="10" language="markdown">
## Summary

Key findings from the data analysis:
- Dataset contains 500 samples with 5 features
- No missing values in this synthetic dataset
- Features are uniformly distributed
- Features show various levels of correlation
</VSCode.Cell>
```
````
