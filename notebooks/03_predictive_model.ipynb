````python
```xml
<!-- filepath: c:\Users\datir\predictive\notebooks\03_predictive_model.ipynb -->
<VSCode.Cell id="1" language="markdown">
# Predictive Model Notebook

This notebook builds, trains, and evaluates predictive machine learning models.
</VSCode.Cell>
<VSCode.Cell id="2" language="python">
# Import Required Libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import sys
import os
import joblib

# Add src directory to path
sys.path.insert(0, os.path.join(os.getcwd(), '..', 'src'))

from synthetic_data import generate_synthetic_dataset
from train_model import train_classification_model, train_regression_model, save_model
from evaluate_model import evaluate_classification, evaluate_regression, print_metrics

# Set visualization style
sns.set_style("whitegrid")
plt.rcParams['figure.figsize'] = (12, 6)
</VSCode.Cell>
<VSCode.Cell id="3" language="markdown">
## Generate Data

Create synthetic data for training and testing.
</VSCode.Cell>
<VSCode.Cell id="4" language="python">
# Generate classification dataset
df = generate_synthetic_dataset(n_samples=800, n_features=8, task='classification', n_classes=2)

# Separate features and target
X = df.drop('target', axis=1).values
y = df['target'].values

print(f"Dataset Shape: {X.shape}")
print(f"Target Shape: {y.shape}")
print(f"Class Distribution: {np.bincount(y)}")
print(f"\nFeature names: {df.columns[:-1].tolist()}")
</VSCode.Cell>
<VSCode.Cell id="5" language="markdown">
## Train Classification Model

Train a Random Forest classifier and evaluate its performance.
</VSCode.Cell>
<VSCode.Cell id="6" language="python">
# Train Random Forest Classifier
results = train_classification_model(
    X, y,
    model_type='random_forest',
    test_size=0.2,
    random_state=42
)

model = results['model']
X_train = results['X_train']
X_test = results['X_test']
y_train = results['y_train']
y_test = results['y_test']

print("Model trained successfully!")
print(f"Training set size: {X_train.shape[0]}")
print(f"Test set size: {X_test.shape[0]}")
</VSCode.Cell>
<VSCode.Cell id="7" language="python">
# Make predictions
y_pred = model.predict(X_test)
y_pred_proba = model.predict_proba(X_test)

# Evaluate model
metrics = evaluate_classification(y_test, y_pred, y_pred_proba)
print_metrics(metrics, task='classification')
</VSCode.Cell>
<VSCode.Cell id="8" language="markdown">
## Feature Importance

Analyze which features are most important for predictions.
</VSCode.Cell>
<VSCode.Cell id="9" language="python">
# Get feature importances
feature_names = [f'feature_{i}' for i in range(X.shape[1])]
importances = model.feature_importances_
indices = np.argsort(importances)[::-1]

# Plot feature importances
plt.figure(figsize=(10, 6))
plt.title("Feature Importance")
plt.bar(range(X.shape[1]), importances[indices], align="center")
plt.xticks(range(X.shape[1]), [feature_names[i] for i in indices], rotation=45)
plt.xlabel("Feature")
plt.ylabel("Importance")
plt.tight_layout()
plt.show()

print("Top 5 Important Features:")
for i in range(min(5, len(indices))):
    print(f"  {feature_names[indices[i]]}: {importances[indices[i]]:.4f}")
</VSCode.Cell>
<VSCode.Cell id="10" language="markdown">
## Model Comparison

Compare different classification models.
</VSCode.Cell>
<VSCode.Cell id="11" language="python">
# Train Logistic Regression for comparison
results_lr = train_classification_model(
    X, y,
    model_type='logistic_regression',
    test_size=0.2,
    random_state=42
)

model_lr = results_lr['model']
y_pred_lr = model_lr.predict(results_lr['X_test'])

# Evaluate Logistic Regression
metrics_lr = evaluate_classification(results_lr['y_test'], y_pred_lr)
print("\n" + "="*50)
print("LOGISTIC REGRESSION METRICS")
print("="*50)
for key, value in metrics_lr.items():
    if key not in ['confusion_matrix', 'classification_report']:
        print(f"{key}: {value:.4f}")
</VSCode.Cell>
<VSCode.Cell id="12" language="python">
# Model Comparison Summary
print("\nMODEL COMPARISON")
print("="*50)
print(f"{'Metric':<20} {'Random Forest':<20} {'Logistic Reg':<20}")
print("-"*60)
print(f"{'Accuracy':<20} {metrics['accuracy']:<20.4f} {metrics_lr['accuracy']:<20.4f}")
print(f"{'Precision':<20} {metrics['precision']:<20.4f} {metrics_lr['precision']:<20.4f}")
print(f"{'Recall':<20} {metrics['recall']:<20.4f} {metrics_lr['recall']:<20.4f}")
print(f"{'F1-Score':<20} {metrics['f1']:<20.4f} {metrics_lr['f1']:<20.4f}")
print("="*50)
</VSCode.Cell>
<VSCode.Cell id="13" language="markdown">
## Save Models

Save the trained models for future use.
</VSCode.Cell>
<VSCode.Cell id="14" language="python">
# Create models directory if it doesn't exist
models_dir = os.path.join(os.getcwd(), '..', 'models')
os.makedirs(models_dir, exist_ok=True)

# Save models
save_model(model, os.path.join(models_dir, 'random_forest_classifier.pkl'))
save_model(model_lr, os.path.join(models_dir, 'logistic_regression_classifier.pkl'))

print("Models saved successfully!")
</VSCode.Cell>
<VSCode.Cell id="15" language="markdown">
## Summary

The Random Forest classifier achieved:
- Accuracy: See metrics above
- Precision: See metrics above
- Recall: See metrics above
- F1-Score: See metrics above

The model has been trained and saved for deployment.
</VSCode.Cell>
```
````
